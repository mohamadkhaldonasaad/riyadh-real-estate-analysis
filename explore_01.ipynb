```python
import pandas as pd
import glob

files = sorted(glob.glob("*.csv"))
print("عدد الملفات:", len(files))

df = pd.concat([pd.read_csv(f, encoding="utf-8-sig") for f in files], ignore_index=True)

df = df.rename(columns={
    'المنطقة': 'region', 'المدينة': 'city', 'المدينة / الحي': 'city_district',
    'الرقم المرجعي للصفقة': 'deal_id', 'تاريخ الصفقة ميلادي': 'date',
    'تاريخ الصفقة هجري': 'date_hijri', 'تصنيف العقار': 'property_type',
    'عدد العقارات': 'property_count', 'السعر': 'price', 'المساحة': 'area'
})

df["price"] = df["price"].str.replace(",", "", regex=False).astype(float)
df["area"]  = df["area"].str.replace(",", "", regex=False).astype(float)

df["date"] = pd.to_datetime(df["date"], format="%Y/%m/%d")
df = df.dropna(subset=["date"])                    # حذف الصفوف الستة الفارغة
df = df.drop_duplicates(subset=["deal_id"])        # ضمان إضافي ضد التكرار

df["year"]    = df["date"].dt.year
df["quarter"] = df["date"].dt.quarter

print(df.shape)
print(df["year"].value_counts().sort_index())
```

```python
print(df["city"].value_counts().head(10))
print()
print(df["property_type"].value_counts())
```

```python
riyadh = df[(df["city"] == "الرياض") & (df["property_type"] == "سكني")].copy()

riyadh["district"] = riyadh["city_district"].str.split("/").str[1].str.strip()

print(riyadh.shape)
print(riyadh["district"].nunique(), "حياً")
print(riyadh["district"].value_counts().head(10))
```

```python
riyadh["price_per_sqm"] = riyadh["price"] / riyadh["area"]

riyadh.groupby("district")["price_per_sqm"].mean().sort_values(ascending=False).head(10)
```

```python
summary = riyadh.groupby("district")["price_per_sqm"].agg(["mean", "median", "count"])
summary = summary.sort_values("mean", ascending=False)
print(summary.head(15))
```

```python
print(riyadh["price_per_sqm"].describe())
print()
print(riyadh["price_per_sqm"].quantile([0.001, 0.01, 0.05, 0.5, 0.95, 0.99, 0.999]))
```

```python
before = len(riyadh)

riyadh = riyadh[riyadh["property_count"] == 1]

low  = riyadh["price_per_sqm"].quantile(0.01)
high = riyadh["price_per_sqm"].quantile(0.99)

riyadh = riyadh[(riyadh["price_per_sqm"] >= low) & (riyadh["price_per_sqm"] <= high)]

print("قبل:", before)
print("بعد:", len(riyadh))
print("حُذف:", before - len(riyadh), "صفاً")
print("النطاق:", round(low), "إلى", round(high))
```

```python
summary = riyadh.groupby("district")["price_per_sqm"].agg(["mean", "median", "count"])
print(summary.sort_values("mean", ascending=False).head(15))
```

```python
counts = riyadh["district"].value_counts()

valid_districts = counts[counts >= 100].index

riyadh = riyadh[riyadh["district"].isin(valid_districts)]

print("عدد الأحياء الصالحة:", len(valid_districts))
print("عدد الصفقات المتبقية:", len(riyadh))
```

```python
riyadh.to_csv("riyadh_clean.csv", index=False, encoding="utf-8-sig")

print("تم الحفظ:", len(riyadh), "صفاً")
```
