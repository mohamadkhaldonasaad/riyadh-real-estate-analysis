```python
import pandas as pd

riyadh = pd.read_csv("riyadh_clean.csv", encoding="utf-8-sig")

print(riyadh.shape)
print(riyadh["district"].nunique(), "حياً")
```

```python
yearly = riyadh.groupby(["district", "year"])["price_per_sqm"].mean().reset_index()

print(yearly.shape)
print(yearly.head(10))
```

```python
pivot = yearly.pivot(index="district", columns="year", values="price_per_sqm")

print(pivot.shape)
print(pivot.head(10))
print()
print("أحياء ناقصة سنة:", pivot.isna().any(axis=1).sum())
```

```python
pivot["change_24_25"] = (pivot[2025] - pivot[2024]) / pivot[2024] * 100

print(pivot[[2024, 2025, "change_24_25"]].sort_values("change_24_25", ascending=False).head(15))
```

```python
check = riyadh[riyadh["district"].isin(["الندوة", "السحاب", "الوسام", "سدرة", "الفاروق"])]

result = check.groupby(["district", "year"]).agg(
    count=("price_per_sqm", "size"),
    median_price=("price_per_sqm", "median"),
    median_area=("area", "median")
)

print(result)
```

```python
stats = riyadh.groupby(["district", "year"]).agg(
    count=("price_per_sqm", "size"),
    median_price=("price_per_sqm", "median"),
    median_area=("area", "median")
).reset_index()

full = stats[stats["year"].isin([2024, 2025])]

pivot = full.pivot(index="district", columns="year",
                   values=["count", "median_price", "median_area"])

print(pivot.shape)
print(pivot.head())
```

```python
riyadh["date"] = pd.to_datetime(riyadh["date"])

cutoff = pd.Timestamp("2025-03-29")

riyadh["period"] = riyadh["date"].apply(lambda d: "before" if d < cutoff else "after")

print(riyadh.groupby("period").size())
```

```python
q = riyadh.groupby(["year", "quarter"]).size().reset_index(name="deals")

q["label"] = q["year"].astype(str) + " Q" + q["quarter"].astype(str)

print(q[["label", "deals"]])
```

```python
by_district = riyadh.groupby(["district", "period"]).size().unstack(fill_value=0)

by_district["retention"] = by_district["after"] / by_district["before"] * 100

print(by_district.sort_values("retention", ascending=False).head(20))
```

```python
price = riyadh.groupby("district")["price_per_sqm"].median()
by_district["median_price"] = price

solid = by_district[by_district["before"] >= 100]

print("=== الأعلى بقاءً ===")
print(solid.sort_values("retention", ascending=False).head(15))
print()
print("=== الأدنى بقاءً ===")
print(solid.sort_values("retention").head(15))
```

```python
solid["area_med"] = riyadh.groupby("district")["area"].median()

top    = solid.sort_values("retention", ascending=False).head(15)
bottom = solid.sort_values("retention").head(15)

print("الأعلى بقاءً — وسيط المساحة:", round(top["area_med"].median()), "م²")
print("الأدنى بقاءً — وسيط المساحة:", round(bottom["area_med"].median()), "م²")
print()
print("الأعلى بقاءً — وسيط السعر:", round(top["median_price"].median()), "ريال/م²")
print("الأدنى بقاءً — وسيط السعر:", round(bottom["median_price"].median()), "ريال/م²")
```

```python
print(riyadh.groupby("period")["price_per_sqm"].median())
print()
print(riyadh.groupby(["year", "quarter"])["price_per_sqm"].median())
```

```python
pb = riyadh[riyadh["period"] == "before"].groupby("district")["price_per_sqm"].median()
pa = riyadh[riyadh["period"] == "after"].groupby("district")["price_per_sqm"].median()

cmp = pd.DataFrame({"before": pb, "after": pa}).dropna()
cmp["change"] = (cmp["after"] - cmp["before"]) / cmp["before"] * 100

print("عدد الأحياء المقارَنة:", len(cmp))
print("وسيط التغير داخل الأحياء:", round(cmp["change"].median(), 1), "%")
print("ارتفعت:", (cmp["change"] > 0).sum(), "حياً")
print("انخفضت:", (cmp["change"] < 0).sum(), "حياً")
```

```python
cmp.to_csv("price_change_by_district.csv", encoding="utf-8-sig")
by_district.to_csv("retention_by_district.csv", encoding="utf-8-sig")
print("تم الحفظ")
```

```python
before = riyadh[riyadh["period"] == "before"]
after  = riyadh[riyadh["period"] == "after"]

summary = pd.DataFrame({
    "deals_before": before.groupby("district").size(),
    "deals_after":  after.groupby("district").size(),
    "price_before": before.groupby("district")["price_per_sqm"].median(),
    "price_after":  after.groupby("district")["price_per_sqm"].median(),
    "median_area":  riyadh.groupby("district")["area"].median(),
})

summary[["deals_before", "deals_after"]] = summary[["deals_before", "deals_after"]].fillna(0)

summary["retention"]    = summary["deals_after"] / summary["deals_before"] * 100
summary["price_change"] = (summary["price_after"] - summary["price_before"]) / summary["price_before"] * 100

summary = summary.round(2).reset_index()
summary.to_csv("district_summary.csv", index=False, encoding="utf-8-sig")

print(summary.shape)
print(summary.head())
```

```python
quarterly = riyadh.groupby(["year", "quarter"]).agg(
    deals=("deal_id", "size"),
    median_price=("price_per_sqm", "median")
).reset_index()

quarterly["label"] = quarterly["year"].astype(str) + " Q" + quarterly["quarter"].astype(str)

quarterly["period"] = quarterly.apply(
    lambda r: "before" if (r["year"] < 2025 or (r["year"] == 2025 and r["quarter"] == 1)) else "after",
    axis=1
)

quarterly = quarterly.round(2)
quarterly.to_csv("quarterly_summary.csv", index=False, encoding="utf-8-sig")

print(quarterly)
```
